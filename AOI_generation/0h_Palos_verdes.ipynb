```python
import geopandas as gpd
from shapely.geometry import box
import matplotlib.pyplot as plt
import pandas as pd
import shapely
from rasterio.crs import CRS
```

```python
geo_pv = shapely.from_wkt('POLYGON((-118.5877 33.6699,-118.1262 33.6699,-118.1262 33.9637,-118.5877 33.9637,-118.5877 33.6699))')
geo_pv
```

```python
df_pv = gpd.GeoDataFrame(geometry=[geo_pv], crs=CRS.from_epsg(4326))
```

```python
df_pv.explore()
```

```python
df_pv.to_file('../AOIs/palos_verdes_ca.geojson', driver='GeoJSON')
```
